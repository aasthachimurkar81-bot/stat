Question 1: A die is rolled. What is the probability of getting:

(a) An even number
A die has outcomes: 1, 2, 3, 4, 5, 6

(a) Probability of getting an even number

Even numbers = 2, 4, 6 → 3 outcomes

$$ P(Even)=\frac{3}{6}=\boxed{\frac12} $$

Answer: 1/2 or 50%

(b) A number greater than 4
Numbers greater than 4 = 5, 6 → 2 outcomes

$$ P(>4)=\frac{2}{6}=\boxed{\frac13} $$

Answer: 1/3 or 33.33%

Question 2: In a class of 50 students:

20 like Mathematics (M)

15 like Science (S)

5 like both subjects

What is the probability that a student chosen at random likes Mathematics or Science?
Formula:

$$ P(M\cup S)=\frac{M+S-Both}{Total} $$ $$ =\frac{20+15-5}{50} $$ $$ =\frac{30}{50}=\boxed{\frac35} $$

Answer: 0.6 or 60%

Question 3: A bag has 3 red and 2 blue balls. If one ball is drawn randomly and is red, what is the probability
that the next ball is also red (without replacement)?
Initially:

Red = 3
Blue = 2
Total = 5

One red ball is already drawn.

So remaining:

Red = 2
Total balls = 4

Therefore,

$$ P(\text{next ball is red})=\frac{2}{4} $$ $$ =\boxed{\frac12} $$

Answer: 1/2 or 50%

Question 4: The population of a school is divided into 60% boys and 40% girls. If you want equal representation
of both genders in the sample, which method should you use: Simple Random Sampling or Stratified Sampling?
Why?
Population:

Boys = 60%
Girls = 40%

We want equal representation of both genders.

Therefore, we should use:

$$ \boxed{\text{Stratified Sampling}} $$

Why?
We divide the population into two groups (boys and girls) and select students from each group separately. This ensures that both genders are represented.

Answer: Stratified Sampling

Question 5: The average height of 1000 students = 160 cm. A sample of 100 students shows an average height
= 158 cm.

Find the sampling error.
Population average height = 160 cm

Sample average height = 158 cm

Formula:

$$ Sampling\ Error = Sample\ Mean - Population\ Mean $$ $$ =158-160 $$ $$ =\boxed{-2\text{ cm}} $$

The magnitude of sampling error is:

$$ \boxed{2\text{ cm}} $$

Answer: 2 cm

Question 6: The population mean salary is ₹50,000 with σ = ₹5,000. If we take a sample of 100 employees, what
is the standard error of the mean (SEM)?
Given:

\(\sigma = ₹5,000\)
\(n=100\)

Formula:

$$ SEM=\frac{\sigma}{\sqrt n} $$

Substitute:

$$ SEM=\frac{5000}{\sqrt{100}} $$ $$ =\frac{5000}{10} $$ $$ =\boxed{₹500} $$

Answer: ₹500

Question 7: In a group of 100 students:

40 like Cricket (C)

30 like Football (F)

10 like both Cricket and Football

Find the probability that a student likes at least one sport?
Given:

Total students = 100
Cricket = 40
Football = 30
Both = 10

Formula:

$$ P(C\cup F)=\frac{C+F-Both}{Total} $$ $$ =\frac{40+30-10}{100} $$ $$ =\frac{60}{100} $$ $$ =\boxed{0.6} $$

Answer: 60%

Question 8: From a deck of 52 cards, two cards are drawn without replacement. What is the probability that
both are Aces?

From a deck of 52 cards, two cards are drawn without replacement. What is the probability that both are Aces?

There are 4 Aces in 52 cards.

Probability of first card being Ace:

$$ \frac{4}{52} $$

After drawing one Ace, there are 3 Aces left and 51 cards left.

Probability of second card being Ace:

$$ \frac{3}{51} $$

Therefore,

$$ P(\text{both Aces})=\frac{4}{52}\times\frac{3}{51} $$ $$ =\frac{12}{2652}=\boxed{\frac{1}{221}} $$

✅ Answer: \(\frac{1}{221}\) ≈ 0.452%

Question 9: A factory produces bulbs with 2% defective rate. If 5 bulbs are chosen at random, what is the
probability that all are non-defective?

A factory produces bulbs with a 2% defective rate. If 5 bulbs are chosen at random, what is the probability that all are non-defective?

Defective rate = 2%

So, non-defective rate:

$$ 100\%-2\%=98\%=0.98 $$

For 5 bulbs to all be non-defective:

$$ P=(0.98)^5 $$ $$ P=0.9039 $$

So,

$$ \boxed{P\approx0.9039} $$

or

$$ \boxed{90.39\%} $$

✅ Answer: 90.39%
